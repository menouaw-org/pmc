# Imports

```python
import random

import math
import matplotlib.pyplot as plt
import numpy as np

from pmc import MLP
```

# Cas de tests, rendu 2

## Classification

### Linear Simple :
        Linear Model : OK
        MLP (2, 1)   : OK

```python
X = np.array([
      [1, 1],
      [2, 3],
      [3, 3]
])
Y = np.array([
      1,
      -1,
      -1
])
```

```python
architecture = [2, 1]
training_steps = 10_000
learning_rate = 0.01
batch_size = 1

targets = np.asarray(Y, dtype=float)
if targets.ndim == 1:
    targets = targets.reshape(-1, 1)

inputs = np.asarray(X, dtype=float)
epochs = math.ceil(training_steps / len(inputs))

model = MLP(architecture, seed=24)
history = model.train(
    inputs,
    targets,
    epochs=epochs,
    learning_rate=learning_rate,
    batch_size=batch_size,
    shuffle=True,
    seed=24,
)

predicted_outputs = np.asarray(model.predict(inputs), dtype=float)

if predicted_outputs.shape[1] == 1:
    predicted_classes = np.where(predicted_outputs[:, 0] >= 0.0, 1, -1)
    expected_classes = targets[:, 0].astype(int)
else:
    predicted_classes = np.argmax(predicted_outputs, axis=1)
    expected_classes = np.argmax(targets, axis=1)

accuracy = np.mean(predicted_classes == expected_classes)

print("expected=", expected_classes)
print("predicted=", predicted_classes)
print(f"accuracy={accuracy:.4f}")
print(f"final_loss={history['loss'][-1]:.6f}")
```

```python
plt.scatter(X[0, 0], X[0, 1], color='blue')
plt.scatter(X[1:3,0], X[1:3,1], color='red')
plt.show()
plt.clf()
```

        MLP (2, 3)      : OK

```python
np.random.seed(24)

X = np.random.random((500, 2)) * 2.0 - 1.0
Y = np.array([[1, -1, -1] if -p[0] - p[1] - 0.5 > 0 and p[1] < 0 and p[0] - p[1] - 0.5 < 0 else
              [-1, 1, -1] if -p[0] - p[1] - 0.5 < 0 and p[1] > 0 and p[0] - p[1] - 0.5 < 0 else
              [-1, -1, 1] if -p[0] - p[1] - 0.5 < 0 and p[1] < 0 and p[0] - p[1] - 0.5 > 0 else
              [-1, -1, -1]for p in X])

X = X[[not np.all(arr == [-1, -1, -1]) for arr in Y]]
Y = Y[[not np.all(arr == [-1, -1, -1]) for arr in Y]]
```
